## 1. Classes and Objects

- A **class** is like a **blueprint** or **template** that includes variables, **attributes** (values) and **methods** (functions).
- An **object** (or **instance**) is a **concrete thing** created from that blueprint.
- This approach helps develop code that is easier to read, use and modify.


Let's create our own simple class.


```python
class Person:
    def __init__(self, name , age): #constructor - special method (dunder)
        self.name = name #name attribute
        self.age = age #age attribute

    def describe(self): #custom method
        print(f"{self.name} is {self.age} years old.")

#Instantiation
person1 = Person("Shrek", 35)

print(type(person1))
print(person1.name, person1.age) #access attributes using "."
person1.describe() #access methods using "."
```

Better practices

```python
class Person:
    def __init__(self, name: str , age: int): #type hints
      '''
      This is the constructor, it takes two arguments: name and age.
      '''
      self.name = name
      self.age = age

    def describe(self) -> str:
      '''
      This method does not take any inputs. It prints a string describing the object (person).
      '''
      print(f"{self.name} is {self.age} years old.")

person1 = Person("Shrek", 35)
person2 = Person("Fiona", 30)

print(person1.name, person1.age)
print(person2.name, person2.age)

person1.describe()
person2.describe()
```

## 2. Core Concepts
### 2.1 Encapsulation

Two main ideas:

1. **Bundling data and methods together** inside a class.
2. **Controlling access** to that data — hiding internal details and exposing a clean interface.

Python doesn’t have true *private* variables, but we use **naming conventions**:

- `public_attribute` → meant to be accessed from outside
- `_protected_attribute` → “internal use” (convention)
- `__private_attribute` → name-mangled to discourage direct access

Let's see an example.


```python
class Person:
    def __init__(self, name: str , age: int, contact: int): #type hints
      '''
      This is the constructor, it takes 3 arguments: name, age and contact.
      '''
      self.name = name
      self.age = age
      self._contact = contact # protected_attribute

    def describe(self):
      '''
      This method does not take any inputs. It prints a string describing the object (person).
      '''
      print(f"{self.name} is {self.age} years old.")

    def get_contact(self): #getter method
      '''
      This method does not take any inputs, but returns a protected attribute - contact.
      '''
      return self._contact

    def update_contact(self, new_contact: int):
      '''
      This method requires an integer input, and updates the protected contact attribute.
      '''
      self._contact = new_contact
      print(f"{self.name}'s updated contact:", self._contact)


person1 = Person(name = "Shrek", age = 35, contact = 1247)

print(person1.name, person1.age, person1.get_contact()) #using method get_contact to retrieve contact attribute

person1.update_contact(new_contact=1112)
```

### 2.2 Abstraction

**Abstraction** is about hiding complexity and exposing a simple interface of a class so that a user doesn't have to know all the internal details. It often uses concepts of encapsulation.

Developers also use the `@abstractmethod` decorator for this. You can learn more about it [here](https://www.geeksforgeeks.org/python/abstract-classes-in-python/)

Let’s build a slightly more abstract example involving bank balances:


```python
class Person:
    def __init__(self, name: str , age: int, balance = 0): #attribute with a default value
      '''
      This is the constructor, it takes multiple arguments now. Some are protected.
      '''
      self.name = name
      self.age = age
      self._balance = balance   # internal balance state (protected)
      self._transactions = []   # internal list of all transactions (protected)

    def describe(self):
      '''
      This method does not take any inputs. It returns a string describing the object (person).
      '''
      print(f"{self.name} is {self.age} years old.")

    def get_balance(self):
        """Public method: return current balance."""
        return self._balance

    # ===== Internal helper function =====

    def _apply_transaction(self, tx_type: str, amount: float):
        """
        - inputs: transaction type and amount
        - updates the balance (in deposit and withdraw functions below)
        - records the transaction in the history
        """
        self._balance += amount
        record = {
            "type": tx_type,
            "amount": amount,
            "balance_after": self._balance,
        }
        self._transactions.append(record)

    def deposit(self, amount: float):
        """Public method:
        input: deposited amount
        output: updates transaction history (and conditional print statement for sanity check)."""
        if amount <= 0:
            print("Deposit amount must be positive.")
            return

        # Update internal state through the helper
        self._apply_transaction("DEPOSIT", amount)
        # print(f"Deposited {amount:.2f}. New balance: {self._balance:.2f}")

    def withdraw(self, amount: float):
        """Public method:
        input: withdrawn amount
        output: updates transaction history (and conditional print statement for sanity check)."""
        if amount <= 0:
            print("Withdrawal amount must be positive.")
            return

        if amount > self._balance:
            print("Not enough funds for this withdrawal.")
            return

        # Update internal state through the helper
        self._apply_transaction("WITHDRAW", -amount)
        # print(f"Withdrew {amount:.2f}. New balance: {self._balance:.2f}")

    def show_statement(self):
        """Public method: prints a table of all transactions and balances."""

        print(f"{self.name}'s statement")

        print("-" * 50)
        for index, tx in enumerate(self._transactions):
            print(f"{index+1}. {tx['type']:8} | {tx['amount']:8.2f} | Balance: {tx['balance_after']:8.2f}")
        print("-" * 50)
        print(f"Final balance: {self._balance:.2f}\n")


person1 = Person("Shrek", 35, 0)

person1.deposit(100)
person1.deposit(50)
person1.withdraw(30)

print("Current balance :", person1.get_balance(), "\n")

person1.show_statement()
```

### 2.3 Inheritance

**Inheritance** lets you create a new class that **reuses** and **extends** an existing one.

- **Parent class** (or base class): the general class
- **Child class** (or derived class): the more specific class

The child class automatically inherits **attributes and methods** of the parent using the `super()` method, and can also:

- Add new methods/attributes
- Override existing methods

Let's explore inheritance!


```python
class Person:
    def __init__(self, name: str , age: int, contact: int): #attribute with a default value
      '''
      This is the constructor, it takes 3 arguments: name, age and contact.
      '''
      self.name = name
      self.age = age
      self._contact = contact

    def get_contact(self):
      '''
      This method does not take any inputs. It returns a protected attribute - contact.
      '''
      print(self._contact)

    def describe(self):
      '''
      This method does not take any inputs. It returns a string describing the object (person).
      '''
      print(f"{self.name} is {self.age} years old.")

class Student(Person):
    def __init__(self, name: str, age: int, contact: int, student_id: int):
        super().__init__(name, age, contact) #super() method helps us inherit the Person class attributes without explicity redefining them
        self.student_id = student_id

class IDLStudent(Student):
    def __init__(self, name: str, age: int, contact:int, student_id:int, grades:int):
        super().__init__(name, age, contact, student_id)
        self._grades = grades

    def get_grades(self):
        '''
        This method does not take any inputs. It returns a protected attribute - grades.
        '''
        return self._grades

    def describe(self): #modified parent method
        '''
        This method does not take any inputs. It returns a string describing an IDL student.
        '''
        print(f"{self.name} is an IDL student!")


person1 = Student(name = "Shrek", age = 35, contact = 1112, student_id = 42)
person2 = IDLStudent(name = "Fiona", age = 30, contact = 1113, student_id = 37, grades = 100)

print(person1.get_contact())
person1.describe()
person2.describe()
```

### 2.4 Polymorphism

**Polymorphism** literally means "many forms" in Greek.

In OOP, it usually means,
different classes can define methods with the **same name**, and you can call that method without caring which class you are dealing with.

Using polymorhism we can put multiple objects instantiated using different classes in a list and call a common method on them that they share. We don't need to know the exact classes that they are instantiated with.

Let's consider the following example:

```python
class Person:
    def __init__(self, name: str , age: int, contact: int): #attribute with a default value
      '''
      This is the constructor, it takes 3 arguments: name, age and contact.
      '''
      self.name = name
      self.age = age
      self._contact = contact

    def get_contact(self):
      '''
      This method does not take any inputs. It returns a protected attribute - contact.
      '''
      return self._contact

    def describe(self):
      '''
      This method does not take any inputs. It returns a string describing the object (person).
      '''
      print(f"{self.name} is {self.age} years old.")

class Student(Person):
    def __init__(self, name: str, age: int, contact: int, student_id: int):
        super().__init__(name, age, contact) #super() method helps us inherit the Person class attributes without explicity redefining them
        self.student_id = student_id

class IDLStudent(Student):
    def __init__(self, name: str, age: int, contact:int, student_id:int, grades:int):
        super().__init__(name, age, contact, student_id)
        self._grades = grades

    def get_grades(self):
        '''
        This method does not take any inputs. It returns a protected attribute - grades.
        '''
        return self._grades

    def describe(self):
        '''
        This method does not take any inputs. It returns a string describing an IDL student.
        '''
        print(f"{self.name} is an IDL student!")


people = [Person("Shrek", 35, 1112), Student("Fiona", 30, 1113, 37), IDLStudent("Ron", 45, 1115, 42, 100)]

for person in people:
  person.describe()
```
